### Anomaly Handling

The `sum_gamerounds` variable is highly right-skewed, with most players recording relatively few rounds and a small number of highly engaged players showing substantially larger values.

Anomaly detection was performed using an SQL-based IQR rule. Players with `sum_gamerounds` above Q3 + 1.5 × IQR were flagged for investigation.

One extreme observation with an exceptionally large number of rounds was identified and treated separately from the remaining high-engagement players. The extreme observation was excluded from the segmentation analysis because it was considered disproportionately influential on the engagement-tier boundaries.

Other high-round observations were retained because they represent plausible high-engagement behaviour and are relevant to the segmentation objective.

The original dataset was preserved, and subsequent segmentation was performed using the cleaned dataset.


```python
segmented_df = con.execute("""
    WITH ranked_players AS (
        SELECT
            userid,
            version,
            sum_gamerounds,
            retention_1,
            retention_7,
            NTILE(3) OVER (
                ORDER BY sum_gamerounds
            ) AS tier_number
        FROM cookie_cats_clean
    )
    SELECT
        userid,
        version,
        sum_gamerounds,
        retention_1,
        retention_7,
        CASE
            WHEN tier_number = 1 THEN 'Low'
            WHEN tier_number = 2 THEN 'Medium'
            WHEN tier_number = 3 THEN 'High'
        END AS engagement_tier
    FROM ranked_players
""").df()

segmented_df.head()
```

### Engagement Segmentation

Players were divided into three approximately equal-sized engagement groups using the SQL `NTILE(3)` window function on `sum_gamerounds`:

- Low engagement
- Medium engagement
- High engagement

The segmentation was performed on the cleaned dataset after excluding the previously identified extreme engagement outlier.

Because `sum_gamerounds` is highly right-skewed, the three tiers contain similar numbers of players but have unequal round-count ranges. The high-engagement tier therefore covers a substantially wider range of game rounds than the lower tiers.

Retention was then summarized by engagement tier and experimental variant (`gate_30` vs `gate_40`) to investigate whether the effect of the gate change varies across player segments.

```python
import numpy as np
import pandas as pd

from scipy.stats import chi2_contingency
from statsmodels.stats.multitest import multipletests
```

```python
def chi_square_test(data, segment, retention_col):

    subset = data[
        data["engagement_tier"] == segment
    ]

    contingency_table = pd.crosstab(
        subset["version"],
        subset[retention_col]
    )

    chi2, p_value, dof, expected = chi2_contingency(
        contingency_table
    )

    return p_value
```

```python
segments = ["Low", "Medium", "High"]

chi_results = []

for segment in segments:

    for retention_col in ["retention_1", "retention_7"]:

        p_value = chi_square_test(
            df_segmented,
            segment,
            retention_col
        )

        chi_results.append({
            "engagement_tier": segment,
            "retention_metric": retention_col,
            "p_value": p_value
        })

chi_results = pd.DataFrame(chi_results)

chi_results
```

```python
def calculate_lift(data, segment, retention_col):

    subset = data[
        data["engagement_tier"] == segment
    ]

    retention_rates = (
        subset.groupby("version")[retention_col]
        .mean()
    )

    gate30_rate = retention_rates["gate_30"]
    gate40_rate = retention_rates["gate_40"]

    lift = gate40_rate - gate30_rate

    return lift
```

```python
lift_results = []

for segment in segments:

    d1_lift = calculate_lift(
        df_segmented,
        segment,
        "retention_1"
    )

    d7_lift = calculate_lift(
        df_segmented,
        segment,
        "retention_7"
    )

    lift_results.append({
        "engagement_tier": segment,
        "d1_lift": d1_lift,
        "d7_lift": d7_lift
    })

lift_results = pd.DataFrame(lift_results)

lift_results
```

```python
lift_results["d1_lift_pp"] = lift_results["d1_lift"] * 100
lift_results["d7_lift_pp"] = lift_results["d7_lift"] * 100

lift_results[
    ["engagement_tier", "d1_lift_pp", "d7_lift_pp"]
]
```

```python
def bootstrap_lift_ci(
    data,
    segment,
    retention_col,
    n_bootstrap=5000,
    random_state=42
):

    subset = data[
        data["engagement_tier"] == segment
    ]

    gate30 = subset[
        subset["version"] == "gate_30"
    ][retention_col].astype(int).to_numpy()

    gate40 = subset[
        subset["version"] == "gate_40"
    ][retention_col].astype(int).to_numpy()

    rng = np.random.default_rng(random_state)

    bootstrap_lifts = []

    for _ in range(n_bootstrap):

        sample30 = rng.choice(
            gate30,
            size=len(gate30),
            replace=True
        )

        sample40 = rng.choice(
            gate40,
            size=len(gate40),
            replace=True
        )

        lift = sample40.mean() - sample30.mean()

        bootstrap_lifts.append(lift)

    lower = np.percentile(
        bootstrap_lifts,
        2.5
    )

    upper = np.percentile(
        bootstrap_lifts,
        97.5
    )

    return lower, upper
```

```python
bootstrap_results = []

for segment in segments:

    for retention_col in ["retention_1", "retention_7"]:

        lower, upper = bootstrap_lift_ci(
            df_segmented,
            segment,
            retention_col
        )

        bootstrap_results.append({
            "engagement_tier": segment,
            "retention_metric": retention_col,
            "ci_lower": lower,
            "ci_upper": upper
        })

bootstrap_results = pd.DataFrame(
    bootstrap_results
)

bootstrap_results
```

```python
bootstrap_results["ci_lower_pp"] = (
    bootstrap_results["ci_lower"] * 100
)

bootstrap_results["ci_upper_pp"] = (
    bootstrap_results["ci_upper"] * 100
)

bootstrap_results[
    [
        "engagement_tier",
        "retention_metric",
        "ci_lower_pp",
        "ci_upper_pp"
    ]
]
```

```python
final_results = []

for segment in segments:

    # -----------------
    # D1
    # -----------------

    d1_lift = calculate_lift(
        df_segmented,
        segment,
        "retention_1"
    )

    d1_lower, d1_upper = bootstrap_lift_ci(
        df_segmented,
        segment,
        "retention_1"
    )

    d1_p = chi_square_test(
        df_segmented,
        segment,
        "retention_1"
    )

    # -----------------
    # D7
    # -----------------

    d7_lift = calculate_lift(
        df_segmented,
        segment,
        "retention_7"
    )

    d7_lower, d7_upper = bootstrap_lift_ci(
        df_segmented,
        segment,
        "retention_7"
    )

    d7_p = chi_square_test(
        df_segmented,
        segment,
        "retention_7"
    )

    final_results.append({
        "engagement_tier": segment,

        "d1_lift_pp": d1_lift * 100,
        "d1_ci_lower_pp": d1_lower * 100,
        "d1_ci_upper_pp": d1_upper * 100,
        "d1_p_value": d1_p,

        "d7_lift_pp": d7_lift * 100,
        "d7_ci_lower_pp": d7_lower * 100,
        "d7_ci_upper_pp": d7_upper * 100,
        "d7_p_value": d7_p
    })

final_results = pd.DataFrame(final_results)

final_results
```

```python
all_pvalues = final_results[
    ["d1_p_value", "d7_p_value"]
].to_numpy().flatten()

adjusted_results = multipletests(
    all_pvalues,
    alpha=0.05,
    method="bonferroni"
)

adjusted_pvalues = adjusted_results[1]

final_results["d1_p_adjusted"] = adjusted_pvalues[::2]
final_results["d7_p_adjusted"] = adjusted_pvalues[1::2]

final_results
```

```python
final_display = final_results[
    [
        "engagement_tier",
        "d1_lift_pp",
        "d1_ci_lower_pp",
        "d1_ci_upper_pp",
        "d1_p_adjusted",
        "d7_lift_pp",
        "d7_ci_lower_pp",
        "d7_ci_upper_pp",
        "d7_p_adjusted"
    ]
].copy()

final_display.round(2)
```

```python
X = df_segmented[
    ["sum_gamerounds", "log_gamerounds"]
].copy()
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
silhouette_results = []

for k in range(2, 7):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=5
    )

    labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels,
        sample_size=10000,
        random_state=42
    )

    silhouette_results.append({
        "k": k,
        "silhouette_score": score
    })

silhouette_results
```

```python
silhouette_df = pd.DataFrame(silhouette_results)

silhouette_df
```

```python
X_train = train_df[
    ["version", "engagement_tier", "version_x_segment"]
]

X_test = test_df[
    ["version", "engagement_tier", "version_x_segment"]
]

y_train = train_df["retention_7_int"]
y_test = test_df["retention_7_int"]
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                drop="first",
                handle_unknown="ignore"
            ),
            ["version", "engagement_tier", "version_x_segment"]
        )
    ]
)

logistic_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

logistic_pipeline.fit(
    X_train,
    y_train
)
```

```python
test_probabilities = logistic_pipeline.predict_proba(
    X_test
)[:, 1]

test_auc = roc_auc_score(
    y_test,
    test_probabilities
)

print(f"Test ROC-AUC: {test_auc:.4f}")
```

```python
train_probabilities = logistic_pipeline.predict_proba(
    X_train
)[:, 1]

train_auc = roc_auc_score(
    y_train,
    train_probabilities
)

print(f"Train ROC-AUC: {train_auc:.4f}")
print(f"Test ROC-AUC:  {test_auc:.4f}")
print(f"Difference:    {train_auc - test_auc:.4f}")
```
